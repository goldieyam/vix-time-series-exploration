```python
#import libraries to project
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

#load data from the web
url = "https://raw.githubusercontent.com/datasets/finance-vix/master/data/vix-daily.csv"
df = pd.read_csv(url)

#data cleaning and preparation
df['DATE'] = pd.to_datetime(df['DATE'])
df = df[['DATE', 'CLOSE']]
df = df.rename(columns={'DATE': 'Date', 'CLOSE': 'Value'})

#visualise moving averages trend
df['Trend'] = df['Value'].rolling(window=7).mean()

plt.plot(df['Date'], df['Value'], label='Actual', alpha=0.5)
plt.plot(df['Date'], df ['Trend'], label='7-Day Trend', color='red')
plt.legend()
plt.title("Trend Analysis")
plt.xlabel("Date")
plt.ylabel("VIX Value")
plt.grid(True)
plt.show()

#create trendline
#convert date to numbers and remove empty rows
df = df.dropna()
df['DateNumber'] = df['Date'].map(pd.Timestamp.toordinal)

#input and output defined
x = df['DateNumber'].values.reshape(-1,1)
y = df['Value'].values

#regression model training
model = LinearRegression()
model.fit(x,y)

#prediction of next day value
import datetime
last_date = df['Date'].max()
tomorrow = last_date + datetime.timedelta(days=1)
tomorrow_num = tomorrow.toordinal()
predicted = model.predict([[tomorrow_num]])

print("Predicted value for tomorrow:", round(predicted[0],2))
```
